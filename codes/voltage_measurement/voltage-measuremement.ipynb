```python
import serial
import openpyxl
from openpyxl import Workbook, load_workbook
from datetime import datetime
import os

PORT = "COM6"      # change to your ESP32's port (Windows: COMx, Linux/Mac: /dev/ttyUSB0 or /dev/cu.usbserial-xxxx)
BAUD = 9600
FILENAME = "voltage.xlsx"

# Load existing file or create a new one
if os.path.exists(FILENAME):
    wb = load_workbook(FILENAME)
    ws = wb.active
else:
    wb = Workbook()
    ws = wb.active
    ws.title = "Voltage Log"
    ws.append(["Timestamp", "Voltage (V)"])
    wb.save(FILENAME)

ser = serial.Serial(PORT, BAUD, timeout=1)
print(f"Listening on {PORT}... Press Ctrl+C to stop.")

try:
    while True:
        line = ser.readline().decode("utf-8", errors="ignore").strip()
        if line:
            try:
                voltage = float(line)
                timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
                ws.append([timestamp, voltage])
                wb.save(FILENAME)  # save after every reading
                print(f"{timestamp} -> {voltage} V")
            except ValueError:
                # ignore lines that aren't valid numbers (e.g. boot messages)
                pass
except KeyboardInterrupt:
    print("\nStopped. Data saved to", FILENAME)
    ser.close()
```
